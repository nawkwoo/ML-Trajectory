```python
# =========================================================
# 1~16.txt 에서 RAW xyz(6번 인덱스)만 추출해 csv로 저장하는 코드
# =========================================================
import pandas as pd
from pathlib import Path

# --- 경로 세팅 ---
base_dir = Path().resolve()
data_dir = base_dir / "data" / "circle"
output_csv = base_dir / "data" / "circle" / "circle_raw_xyz.csv"

# --- xyz 추출 함수 ---
def load_raw_xyz(idx):
    file_path = data_dir / f"{idx}.txt"
    xs, ys, zs = [], [], []

    with open(file_path, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue

            cols = line.split(",")

            # 6번 인덱스가 있는지 확인
            if len(cols) <= 6:
                continue

            xyz_str = cols[6].strip()
            parts = xyz_str.split("/")

            if len(parts) != 3:
                continue

            try:
                x, y, z = map(float, parts)
            except:
                continue

            xs.append(x)
            ys.append(y)
            zs.append(z)

    return xs, ys, zs

# --- 전체 raw 저장 ---
rows = []

for idx in range(1, 17):
    xs, ys, zs = load_raw_xyz(idx)
    T = len(xs)

    for t in range(T):
        rows.append({
            "file_id": idx,
            "t": t,
            "x": xs[t],
            "y": ys[t],
            "z": zs[t],
        })

df_raw = pd.DataFrame(rows)

# --- CSV로 저장 ---
df_raw.to_csv(output_csv, index=False)

print("RAW xyz CSV 저장 완료:", output_csv)
df_raw.head()
```
